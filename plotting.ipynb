## Stride Scheduling (8:4:2:1) Ticket allocation

```python
import matplotlib.pyplot as plt
import numpy as np

# Data from the user
time_quanta = np.array([20, 40, 60, 80, 100])

# Actual data (Avg)
actual = {
    'A': np.array([10, 21.2, 32, 42, 53]),
    'B': np.array([5, 10.8, 16, 21, 27]),
    'C': np.array([3, 5, 8, 11, 13]),
    'D': np.array([2, 3, 4, 6, 7])
}

# Expected data
expected = {
    'A': np.array([10, 20, 30, 40, 50]),
    'B': np.array([5, 10, 15, 20, 25]),
    'C': np.array([3, 6, 9, 12, 15]),
    'D': np.array([2, 4, 6, 8, 10])
}

colors = {'A': 'red', 'B': 'green', 'C': 'blue', 'D': 'orange'}

plt.figure(figsize=(8,6))

# Plot actual data as solid lines with labels (just A, B, C, D)
for client, vals in actual.items():
    plt.plot(time_quanta, vals, color=colors[client], label=client, marker='o')

# Plot expected data as dashed gray lines without labels
for vals in expected.values():
    plt.plot(time_quanta, vals, color='gray', linestyle='dashed')

plt.xlabel('Time (quanta)')
plt.ylabel('Cumulative Quanta')
plt.title('Stride Scheduling (8:4:2:1) Ticket allocation')
plt.legend()
plt.grid(True)
plt.tight_layout()
plt.show()
```
